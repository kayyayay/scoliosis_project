## 1. Maximum-angle reliability

The largest curve angle in each image provides one case-level value per observer. The comparison uses the same **498 complete radiographs** for both pathways.

| Pathway | ICC(A,1) | 95% bootstrap CI | Mean human pairwise difference |
|---|---:|---:|---:|
| Human-centroid spline | 0.917 | 0.898–0.934 | 2.64° |
| Traditional manual Cobb | 0.867 | 0.841–0.886 | 3.38° |

![Maximum-angle ICC comparison](../outputs/figures/human_maximum_angle_icc.png)

The higher spline ICC indicates that the observers were more consistent under the spline pathway for the maximum recorded angle.

## 2. Paired within-image disagreement

For every image and pathway, the six absolute differences among four observers were averaged. Manual disagreement was **0.74° higher** than spline disagreement on average (bootstrap 95% CI: **0.54°–0.94°**). Spline disagreement was lower in **63.7%** of images.

![Per-image observer disagreement](../outputs/figures/human_per_image_disagreement.png)

## 3. Matched-curve agreement

For anatomically matched curves, spline MAE was **2.42°** compared with **3.39°** for manual measurement. Spline pairs were within 5° in **91.8%** of comparisons, compared with **80.7%** for manual pairs.

Both the case-level and curve-level results support greater interobserver reliability for spline-based measurement. They do not show which pathway is objectively accurate because the dataset contains no unquestionable ground truth.
