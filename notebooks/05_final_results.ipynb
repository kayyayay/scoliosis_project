## Answers to the research questions

1. **Human reliability:** spline-based maximum-angle ICC was 0.917 versus 0.867 for traditional manual measurement.
2. **Neural-network agreement:** maximum-angle MAE was 1.63° and CCC was 0.950 against the median human spline consensus.
3. **Curve structure:** approximately 86%–87% curve coverage was achieved at the primary overlap threshold; not every recorded curve could be confidently paired.
4. **Endpoint effects:** larger endpoint differences were associated with larger angle differences, but the overall relationship was weak.
5. **Curve magnitude and prediction:** magnitude contributed to exploratory model behavior, but grouped cross-validated error prediction remained weak.

## Main conclusion

Spline-based angle calculation was more reproducible among the four human observers than traditional manual measurement in this released dataset. Neural-network-assisted spline measurements agreed strongly with a median human spline reference, but occasional large differences and the absence of independent external validation prevent a clinical accuracy claim.

## What this project does not claim

- It does not identify an unquestionable ground-truth Cobb angle.
- It does not independently validate the neural network on a new clinical dataset.
- It does not diagnose scoliosis or recommend treatment.
- It does not prove that the 0.50-IoU curve-matching rule is clinically optimal.

The strongest description is: **a reproducible independent analysis of measurement reliability and agreement using a publicly released scoliosis dataset.**
