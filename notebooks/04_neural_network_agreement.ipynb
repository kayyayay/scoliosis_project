## 1. Maximum angle versus human consensus

The human reference is the median of the four observers' maximum spline angles. Among **485 complete radiographs**:

- MAE: **1.63°** (bootstrap 95% CI: 1.43°–1.84°)
- median absolute difference: **1.00°**
- within 3°: **90.9%**
- within 5°: **94.8%**
- CCC: **0.950**
- bias, neural network minus consensus: **−0.59°**

![Neural-network maximum angle versus human consensus](../outputs/figures/neural_vs_human_consensus.png)

## 2. Bland–Altman analysis

![Bland-Altman plot](../outputs/figures/neural_consensus_bland_altman.png)

The 95% limits of agreement were **−5.97° to 4.80°**. Strong overall concordance therefore coexists with occasional substantial individual-case errors.

## 3. Endpoint disagreement

![Angle disagreement by endpoint disagreement](../outputs/figures/endpoint_disagreement.png)

Mean absolute angle disagreement increased from **1.68°** for identical endpoints to **5.71°** when the total endpoint difference was at least five indices. However, the overall Spearman correlation was only **0.155**, so the relationship is positive but weak.

## 4. Exploratory error models

![Grouped cross-validated model performance](../outputs/figures/error_model_performance.png)

Five-fold grouped cross-validation kept all curves from one image in the same fold. Neither linear regression nor random forest materially improved MAE over the training-mean baseline. Endpoint differences and curve magnitude provide some information, but they cannot reliably predict the exact error for an individual curve.
